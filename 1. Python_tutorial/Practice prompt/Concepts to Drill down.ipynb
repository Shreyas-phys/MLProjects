### **Python list behavior**

* Slicing → **copy**
* Assignment like `a = b` → **aliasing**, not copying
* There are **no views** in Python lists
* Mutating either alias mutates the same underlying list

### **NumPy behavior**

* Slicing → **view** (usually)
* Assignment like `a = b` → alias (same as Python)

You mixed NumPy slicing semantics with Python lists.
This is extremely common — and now corrected.


# 🔥 **Shallow Copy vs Deep Copy (Python List Version)**

## **1. Shallow Copy**

A **shallow copy** creates a new list,
but **does NOT copy the inner objects**.

Example:

```python
lst  = [[1,2], [3,4]]
lst2 = lst[:]   # shallow copy
```

Visualization:

```
lst -------> [ [1,2],  [3,4] ]
               ↑        ↑
lst2 -------> [  same   same ]
```

### ✔ What this means:

* `lst` and `lst2` are different outer lists
* **BUT they share the same inner lists**

So:

```python
lst2[0][0] = "X"
```

changes **both** lists, because `[1,2]` is shared.

---

## **2. Deep Copy**

A **deep copy** duplicates everything**, including nested objects.

```python
import copy
lst2 = copy.deepcopy(lst)
```

Visualization:

```
lst  -------> [ [1,2], [3,4] ]
lst2 -------> [ [1,2], [3,4] ]   # totally new copies
```

### ✔ What this means:

Mutating anything inside `lst2` does **NOT** affect `lst`.

# 🧠 Summary Table

| Operation            | Type         | Inner Objects Copied? | Mutations Shared? |
| -------------------- | ------------ | --------------------- | ----------------- |
| `lst2 = lst`         | alias        | no                    | yes               |
| `lst2 = lst.copy()`  | shallow copy | no                    | yes               |
| `lst2 = lst[:]`      | shallow copy | no                    | yes               |
| `lst2 = list(lst)`   | shallow copy | no                    | yes               |
| `copy.deepcopy(lst)` | deep copy    | yes                   | no                |
