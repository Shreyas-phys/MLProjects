### Problem 5 (NumPy Foundations: 2D Matrix Creation & Matrix-Vector Multiplication)

Define the Pauli-$Y$ matrix:


$$\sigma_y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}$$


as a $2 \times 2$ NumPy array `Y` with `dtype=np.complex128`.

Then, given the state vector:


$$\vert{}\psi\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$


as a 1D NumPy array `psi` of shape `(2,)` with `dtype=np.complex128`, compute the action of the operator on the state:


$$\vert{}\phi\rangle = \sigma_y \vert{}\psi\rangle$$


using NumPy's matrix multiplication operator (`@`), and store the result in `phi`.

*Paste your solution (3–4 lines) below.*

```python
Y = np.array([[0,-1j],[1j,0]], dtype=np.complex128)

print(type(Y))
print(Y)

psiT = np.array([[1,0]], dtype="complex128")
psi=psiT.T
print(psi.shape)
print(psi)

phi = Y@psi

print(phi)

```
