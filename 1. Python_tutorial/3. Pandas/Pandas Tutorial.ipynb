````markdown
## Core Pandas Attributes — Short Notes

### `.values`
Returns the underlying data as a NumPy array.
- Drops row/column labels
- Shape = (rows, columns)
- Useful for plotting and linear algebra
- Slightly legacy; works but not explicit

Use when you want *just the numbers*.

---

### `.to_numpy()`
Explicit and preferred version of `.values`.
- Same output as `.values`
- Safer with newer pandas versions
- Makes intent clear: “convert to NumPy”

Use this by default instead of `.values`.

---

### `.columns`
Holds the column labels.
- Often strings or numbers
- Used for axis labels and iteration

Example:
```python
plt.xticks(range(len(df.columns)), df.columns)
````

---

### `.index`

Holds the row labels.

* Can represent time, site index, IDs, etc.
* Appears on the y-axis in plots

Think of it as the “row coordinate”.

---

### `.shape`

Gives `(number_of_rows, number_of_columns)`.

* Fast sanity check after loading or filtering
* Helps catch wrong slicing early

---

### `.dtypes`

Shows the data type of each column.

* Critical for debugging silent issues
* Mixed dtypes often cause slowdowns or plot errors

Always check when things behave strangely.

---

### `.T`

Transpose of the DataFrame.

* Rows ↔ columns
* Preserves labels (unlike NumPy transpose)

Very common when switching between
“site × time” and “time × site” views.

---

## Big Picture

* Pandas = labeled data (safe, readable)
* NumPy = raw numbers (fast, math-friendly)

Workflow:

```python
df  → inspect / label / group
 → df.to_numpy()
 → math / plots
```

If you remember nothing else:
**use DataFrame for thinking, NumPy for computing.**
