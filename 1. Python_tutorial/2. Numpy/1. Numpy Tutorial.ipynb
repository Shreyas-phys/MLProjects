## **1.2 Creating Arrays**

### **From Python lists**

```python
a = np.array([1, 2, 3])             # Create a 1D array
b = np.array([[1, 2], [3, 4]])      # Create a 2D array (matrix)
```

`np.array()` converts a Python list (slow, untyped) into a fast, typed NumPy array.

---

### **Specifying a dtype**

```python
c = np.array([1, 2, 3], dtype=np.float32)   # Array with float32 precision
```

Dtypes determine precision, memory usage, and sometimes speed.

## **1.3 Inspecting Array Properties**

### **Understanding shape, dimensions, size, dtype**

```python
a.shape      # Dimensions of the array (e.g., (3,), (2,3))
a.ndim       # Number of dimensions (1D, 2D, 3D...)
a.size       # Total number of elements
a.dtype      # Data type of elements
```

These four attributes are the *core diagnostic tools* for understanding any array.

## **1.4 Changing Data Types**

```python
a.astype(np.float64)     # Convert array to float64
a.astype(int)            # Convert array to int
```

`astype()` always returns a **copy** of the array with the new dtype.

---

## **1.5 Copy vs View (Very Important)**

### **Views (share memory)**

```python
x = np.array([10, 20, 30])
y = x[1:]                  # y is a *view* of x, no new memory
y[0] = 999                 # This changes x as well
```

Slicing returns a **view**, meaning changes propagate to the original.

---

### **Copies (independent memory)**

```python
z = x.copy()               # Create an independent copy
z[0] = 123                 # Does NOT affect x
```

Use `.copy()` whenever you need isolation from mutations.

---

```python
a + 3              # Add 3 to every element
a * 2              # Multiply each element by 2
a ** 2             # Square each element
np.sqrt(a)         # Elementwise square root
np.exp(a)          # Elementwise exponential
```

NumPy applies operations elementwise using fast compiled code.

## **1.7 Why NumPy Is Fast (interview-ready sentence)**

> **NumPy is fast because arrays are stored contiguously in memory and operations run in optimized C loops instead of Python loops.**

This is *exactly* the response expected in ML and software interviews.

## **3.1 Basic 1D Indexing**

```python
a = np.array([10, 20, 30, 40])
a[0]         # First element → 10
a[2]         # Third element → 30
a[-1]        # Last element  → 40
```

Standard Python indexing rules apply.

## **3.2 Basic 1D Slicing**

```python
a[1:3]       # Elements at indices 1 and 2 → [20, 30]
a[:2]        # First two elements → [10, 20]
a[2:]        # From index 2 to end → [30, 40]
a[::2]       # Step of 2 → [10, 30]
a[::-1]      # Reverse → [40, 30, 20, 10]
```

Slicing creates **views**, not copies.

---

## **3.3 2D Indexing**

```python
A = np.array([[1,2,3],
              [4,5,6]])

A[0, 0]      # Row 0, Col 0 → 1
A[1, 2]      # Row 1, Col 2 → 6
```

Use `A[row, column]`.

---

## **3.4 Extracting Entire Rows or Columns**

### **Row extraction**

```python
A[0, :]      # Entire first row → [1, 2, 3]
```

### **Column extraction**

```python
A[:, 1]      # Entire second column → [2, 5]
```

Using `:` selects all elements along that axis.

---

## **3.5 2D Slicing**

```python
A[0:2, 1:3]   # Rows 0–1, Cols 1–2 → [[2,3],
                                    #  [5,6]]
```

Think of slicing as selecting **row-range**, **column-range**.

---

## **3.6 Slicing Produces Views (Not Copies)**

```python
B = A[:, :2]   # View of A
B[0, 0] = 999  # This modifies A as well!
```

Slicing never allocates new memory unless you call `.copy()`.

---

## **3.7 Copy Explicitly When Needed**

```python
C = A[:, :2].copy()   # Independent copy
```

Modifying `C` will not affect `A`.

[Home](#home)


## **4.1 What Is Broadcasting?**
Broadcasting allows NumPy to perform operations on arrays of **different shapes** by automatically expanding one array without copying data.

Example:

```python
a = np.array([1, 2, 3])
a + 5                     # → [6, 7, 8]
```

NumPy "broadcasts" the scalar `5` across all elements.

---

## **4.2 Broadcasting with 1D Arrays**

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

a + b                     # Elementwise addition → [11, 22, 33]
a * b                     # Elementwise multiply → [10, 40, 90]
```

Works only if shapes are compatible.

---

## **4.3 Broadcasting with 2D and 1D Arrays**

```python
A = np.array([[1,2,3],
              [4,5,6]])

v = np.array([10, 20, 30])

A + v                     # v is added to each row of A
```

NumPy expands `v` to match the rows — no data duplication.

---

## **4.4 Broadcasting Rules (Memorize These)**

Two dimensions are compatible if:

1. They are equal
2. OR one of them is `1`

Example:

```
A shape: (4, 3)
v shape: (3,)
→ v becomes (1, 3) then (4, 3)
```

Example of incompatible shapes:

```
(4, 3) and (2,)
```

Cannot broadcast because `3 ≠ 2` and neither dimension is `1`.

---

## **4.5 Expanding Dimensions Manually**

Use `None` or `np.newaxis` to add a dimension.

### **Make a column vector**

```python
a = np.array([1, 2, 3])
a[:, None]                # Shape (3,) → (3,1)
```

### **Make a row vector**

```python
a[None, :]                # Shape (3,) → (1,3)
```

This ensures shapes match when broadcasting.

---

## **4.6 Column-wise and Row-wise Operations**

### **Add a column vector to each column**

```python
A + a[:, None]
```

Shapes:

```
A: (m, n)
a[:, None]: (m, 1)
→ broadcast to (m, n)
```

### **Add a row vector to each row**

```python
A + a[None, :]
```

---

## **4.7 Combining Broadcasting with Basic Ops**

```python
A * 2             # Scalar broadcast
A + v             # Row broadcast
A / A.sum(axis=1, keepdims=True)   # Normalize each row
```

Broadcasting underlies many array operations.

---

## **CHAPTER 5 — Vectorization**

[Home](#home)

## **5.1 What Is Vectorization?**

Vectorization means performing operations on **entire arrays at once**, instead of writing Python loops.

Example:

```python
a = np.array([1, 2, 3])
a * 2                     # Multiply all elements at once
```

Vectorized operations run in compiled C code → **much faster** than Python loops.

---

## **5.2 Replacing Python Loops with NumPy Operations**

### **Slow Python loop**

```python
x = [1, 2, 3]
y = []
for val in x:
    y.append(val * 2)
```

### **Fast NumPy version**

```python
x = np.array([1, 2, 3])
y = x * 2
```

One line, faster, and more expressive.

---

## **5.3 Common Vectorized Elementwise Operations**

```python
a + 3              # Add scalar to all elements
a - 1              # Subtract scalar
a * 5              # Multiply all elements
a / 2              # Divide all elements
a ** 2             # Square all elements
np.sqrt(a)         # Square root
np.exp(a)          # Exponential
np.log(a)          # Natural log
```

All of these operate on **every element automatically**.

---

## **5.4 Vectorized Comparisons**

```python
a > 0              # Elementwise comparison → boolean array
a == 3             # Check equality for each element
(a % 2) == 0       # Check even numbers
```

The output is an array of booleans — no loops needed.

---

## **5.5 Vectorized Arithmetic Between Arrays**

### **Elementwise array operations**

```python
a + b              # Add elementwise
a * b              # Multiply elementwise
a / b              # Divide elementwise
```

Works only when shapes are compatible.

---

## **5.6 Dot Product and Matrix Multiplication**

```python
np.dot(a, b)       # Dot product (1D vectors)
A @ B              # Matrix multiplication
np.matmul(A, B)    # Same as @
```

Vectorized linear algebra using optimized BLAS routines.

---

## **5.7 Avoiding Loops: Typical Transformations**

### **Python loop (slow)**

```python
result = []
for x in a:
    result.append(x**2 + 2*x + 1)
```

### **Vectorized version**

```python
result = a**2 + 2*a + 1
```

Push operations into NumPy — never manually loop over array elements.

---

## **5.8 Broadcasting + Vectorization (Implicit)**

Although broadcasting was covered earlier, most vectorized operations rely on it.

Example (concept only, no deep coverage here):

```python
A + a              # Adds vector a to each row of matrix A
```

[Home](#home)

## **8.1 What Is a Boolean Mask?**

A boolean mask is an array of `True`/`False` values used to **select** or **modify** elements in another array.

Example:

```python
a = np.array([1, -3, 5, -2])
mask = a > 0                    # → [True, False, True, False]
```

Masks allow vectorized filtering **without loops**.

---

## **8.2 Using Masks to Filter Arrays**

```python
a[a > 0]                        # Select only positive values
a[a < 3]                        # Select values < 3
a[a % 2 == 0]                   # Select even numbers
```

Result is a **new 1D array** containing only elements where the mask is True.

---

## **8.3 Masks with 2D Arrays**

```python
A = np.array([[1, 5, -2],
              [7, -1,  3]])

A[A > 0]                        # Returns all positive values (flattened)
```

Mask selection always returns a **1D array**.

---

## **8.4 Combining Masks**

Use bitwise operators:

```python
mask = (a > 0) & (a < 10)       # AND
mask = (a < 0) | (a > 5)        # OR
mask = ~(a == 3)                # NOT
```

Parentheses are required because `&` and `|` have higher precedence.

---

## **8.5 Modifying Arrays Using Masks**

```python
a[a < 0] = 0                    # Replace negative values with 0
A[A < 0] = -1                   # Replace all negatives with -1
```

Masking can be used for fast data cleaning.

---

## **8.6 Counting Elements with Masks**

```python
(a > 0).sum()                   # Count positive values
(A == 0).sum()                  # Count zeros in a matrix
```

Boolean `True` is treated as `1`, `False` as `0`.

---

## **8.7 Masks Based on Aggregates**

```python
threshold = a.mean()
a[a > threshold]                # Values above average
```

Masks can be created using any numeric condition.

---

[Home](#home)

## **9.1 What Is Fancy Indexing?**

Fancy indexing means selecting elements, rows, or columns using **lists or arrays of indices**, instead of slices.

Example:

```python
a = np.array([10, 20, 30, 40, 50])
a[[0, 2, 4]]                      # → [10, 30, 50]
```

It **always returns a copy**, not a view.

---

## **9.2 Selecting Multiple Elements from a 1D Array**

```python
a[[1, 3, 4]]                      # Select elements at indices 1, 3, 4
a[[0, -1]]                        # First and last elements
```

Indices must be a list or NumPy array.

---

## **9.3 Selecting Rows from a 2D Array**

```python
A = np.array([[10,11,12],
              [20,21,22],
              [30,31,32],
              [40,41,42]])

A[[0, 2]]                          # Select rows 0 and 2
```

Produces a new array containing only those rows.

---

## **9.4 Selecting Columns from a 2D Array**

Fancy indexing applies independently to each axis:

```python
A[:, [0, 2]]                       # Select columns 0 and 2
```

Slicing (`:`) on rows, fancy indexing on columns.

---

## **9.5 Selecting Arbitrary Elements (Two Arrays of Indices)**

```python
rows = np.array([0, 1, 2])
cols = np.array([1, 1, 1])

A[rows, cols]                     # Picks A[0,1], A[1,1], A[2,1]
```

Allows you to choose **specific coordinates** element by element.

---

## **9.6 Fancy Indexing Produces a Copy**

```python
sub = A[[0, 2]]
sub[0, 0] = 999
```

The original `A` is unchanged — fancy indexing **never** returns a view.

This differs from slicing, which returns a **view**.

---

## **9.7 Combining Fancy Indexing and Slicing**

```python
A[[0, 2], :]                      # Select rows 0 and 2, all columns
A[:, [1, 2]]                      # Select last two columns
```

You can use slicing on one axis and fancy indexing on the other.

---

## **14.1 Loss Functions**

### **Mean Squared Error (MSE)**

```python
def mse(y, y_pred):
    return np.mean((y - y_pred)**2)
```

Most common loss for regression.

---

### **Mean Absolute Error (MAE)**

```python
def mae(y, y_pred):
    return np.mean(np.abs(y - y_pred))
```

Robust to outliers.

---

### **Binary Cross-Entropy Loss**

Assumes `y` and `y_pred` are in ([0,1]):

```python
def binary_cross_entropy(y, y_pred):
    eps = 1e-9
    return -np.mean(y*np.log(y_pred+eps) + (1-y)*np.log(1-y_pred+eps))
```

Used in logistic regression & binary classifiers.

---

## **14.2 Activation Functions**

### **Sigmoid**

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

---

### **ReLU**

```python
def relu(x):
    return np.maximum(0, x)
```

---

### **Softmax**

Row-wise softmax (batch inputs):

```python
def softmax(x):
    e = np.exp(x - np.max(x, axis=1, keepdims=True))
    return e / e.sum(axis=1, keepdims=True)
```

Subtracting max improves numerical stability.

---

## **14.3 Metrics**

### **Accuracy**

```python
def accuracy(y, y_pred):
    return np.mean(y == y_pred)
```

---

### **Root Mean Squared Error (RMSE)**

```python
def rmse(y, y_pred):
    return np.sqrt(np.mean((y - y_pred)**2))
```

---

### **Simple Confusion Matrix**

Assume binary labels `{0,1}`:

```python
def confusion_matrix(y, y_pred):
    tp = np.sum((y==1) & (y_pred==1))
    tn = np.sum((y==0) & (y_pred==0))
    fp = np.sum((y==0) & (y_pred==1))
    fn = np.sum((y==1) & (y_pred==0))
    return np.array([[tn, fp],
                     [fn, tp]])
```

---

## **14.4 Linear Regression (Closed-Form Solution)**

Solves the normal equation:

$$
w = (X^T X)^{-1} X^T y
$$

```python
def linear_regression_closed_form(X, y):
    return np.linalg.inv(X.T @ X) @ X.T @ y
```

(This assumes (X^T X) is invertible.)

---

## **14.5 Gradient Descent Update Rule**

For weights (w) in linear regression:

$$
w := w - \eta \frac{2}{n} X^T (Xw - y)
$$

```python
def gradient_descent_step(X, y, w, lr=0.01):
    n = len(y)
    grad = (2/n) * X.T @ (X @ w - y)
    return w - lr * grad
```

Works for any differentiable loss.


## **CHAPTER 15 — Performance (How NumPy Works Under the Hood)**

[Home](#home)

### **15.1 The Core Reason NumPy Is Fast**

NumPy is fast because:

1. **It stores numbers in contiguous memory (like C arrays)**
2. **Operations run in compiled C code, not Python**
3. **It uses SIMD instructions → CPU processes many numbers at once**
4. **It avoids Python object overhead**

Interview version:

> **NumPy is fast because it performs vectorized operations in highly optimized C code on contiguous memory, using low-level CPU instructions like SIMD.**

---

### **15.2 Python’s Memory Layout (Why Pure Python Is Slow)**

A Python list:

```python
[1, 2, 3, 4]
```

Internally looks like:

```
[ pointer → PyObject(1),
  pointer → PyObject(2),
  pointer → PyObject(3),
  pointer → PyObject(4) ]
```

Each element is:

* A separate Python object
* Located somewhere else in memory
* Contains metadata (type info, reference count, etc.)

Result:
❌ Bad memory locality
❌ Slow looping
❌ Heavy overhead for each element

---

### **15.3 NumPy’s Memory Layout (Why It’s Fast)**

A NumPy array:

```python
np.array([1, 2, 3, 4])
```

Internally:

```
1 2 3 4 (raw bytes, tightly packed in memory)
```

Exactly like a C array:

```c
int a[] = {1,2,3,4};
```

Benefits:

✔ No object overhead
✔ Excellent memory locality
✔ Perfect for SIMD and vectorized loops
✔ Minimal pointer chasing

---

### **15.4 What Is SIMD? (Beginner-Friendly Explanation)**

**SIMD = Single Instruction, Multiple Data**

Meaning:

> The CPU can perform **one operation on multiple numbers at the same time**.

For example, instead of:

```
multiply 1 by 2  
multiply 2 by 2  
multiply 3 by 2  
multiply 4 by 2
```

The CPU can do:

```
multiply {1,2,3,4} by 2 in one go
```

NumPy uses these instructions automatically inside its C loops.
You don’t write SIMD — NumPy does it for you.

---

### **15.5 NumPy vs Python vs C (Clear Comparison)**

### **Python loop**

```python
for i in range(n):
    a[i] = a[i] * 2
```

* Visits each element one at a time
* Interpreted
* Slow

---

### **C loop**

```c
for (int i = 0; i < n; i++) {
    a[i] *= 2;
}
```

* Much faster
* Tight compiled loop
* Operates directly on contiguous memory

---

### **NumPy**

```python
a *= 2
```

Internally equivalent to optimized C like:

```c
for (i += SIMD_width) {
    load vector of N values
    multiply all values by 2
    store result
}
```

So NumPy often matches or exceeds naïve C performance.

---

### **15.6 Views vs Copies (Performance Implications)**

### **Views (fast)**

Created by slicing, reshaping:

```python
v = a[100:200]   # view → does NOT copy data
```

* O(1)
* No new memory allocated
* Simply adjusts pointer + shape metadata

---

### **Copies (slower)**

Created by fancy indexing or `.copy()`:

```python
c = a[[1,5,7]]   # copy
```

* Allocates new memory
* Copies data element-by-element
* More expensive, especially for large arrays

---

### **15.7 When NumPy Is *Not* Fast**

NumPy is slower if:

### ❌ You use Python loops

```python
for i in range(len(a)):
    a[i] = a[i] * 2
```

### ❌ You use object dtype

```python
np.array(["hello", 1, {}], dtype=object)
```

### ❌ The array is non-contiguous (strided)

```python
b = a[::2]   # skipping elements breaks SIMD efficiency
```

### ❌ Arrays are extremely small (< 100 elements)

Overhead of dispatch dominates.

### ❌ You mix Python and NumPy operations inefficiently

Example:

```python
for x in a:
    np.sqrt(x)
```

Always push work into **vectorized NumPy calls**, not loops.

---

# ### **15.8 In-Place Operations Are Faster**

```python
a += 1
a *= 3
```

Faster because:

* No temporary array is created
* No extra memory is allocated
* CPU loads each element once

Compare to:

```python
a = a + 1   # slower, creates a temporary array
```

For matrices:

```python
C = A @ B
```

If A and B are (n × n):

* Each element of result requires n multiplications
* There are n² elements
* Total operations ≈ n³

Matrix multiplication dominates ML workloads.

---

### **Views use O(1) memory**

Broadcasting checks shape compatibility in constant time.

## **17.1 — Shape Mismatch Errors**

These happen whenever two arrays have incompatible shapes.

---

### **Error: ValueError: operands could not be broadcast together**

```python
a = np.ones((3,3))
b = np.ones((4,))
a + b
```

**Why it happens:**
(3,3) cannot broadcast with (4,).
Broadcasting requires trailing dimensions to match or be 1.

**Fixes:**

```python
b = b[:3]           # match size
b = b.reshape(1,3)  # reshape for row broadcast
b = b[:,None]       # reshape for column broadcast
```

---

### **Error: ValueError: shapes (m,n) and (p,q) not aligned**

Matrix multiplication example:

```python
A @ B
```

**Why:**
Inner dimensions must match: (m × n) @ (n × p)

**Fix:**
Check `A.shape` and `B.shape`
Transpose if appropriate:

```python
B = B.T
```

## **17.2 — Indexing & Slicing Errors**

---

### **Error: IndexError: index X is out of bounds**

```python
a = np.array([10,20,30])
a[3]   # invalid index
```

**Why:**
Valid indices are 0,1,2 → index 3 does not exist.

**Fix:**
Check `a.shape`, correct your loop bounds, or use:

```python
a[-1]   # last element
```

### **Error: too many indices for array**

```python
a = np.array([1,2,3])
a[0,1]    # trying 2D indexing on a 1D array
```

## **17.3 — Type Errors & dtype Problems**

---

### **Error: TypeError: ufunc 'add' did not contain a loop with signature…**

## **17.4 — Copy vs View Errors (Silent Bugs)**

These do *not* raise exceptions — they corrupt your data silently

### **Silent Bug: modifying a slice modifies original array**

```python
a = np.array([1,2,3])
v = a[1:]
v[0] = 99
a   # becomes [1, 99, 3]
```

Fix with `.copy()` when needed.

## **17.6 — Common Linear Algebra Errors**

---

### **Error: LinAlgError: Singular matrix**

```python
np.linalg.inv(A)
```

**Why:**
Matrix is not invertible.

**Fix:**
Use pseudoinverse:

```python
np.linalg.pinv(A)
```

or regularization:

```python
A += 1e-6 * np.eye(A.shape[0])
```

---

### **Error: LinAlgError: Last 2 dimensions must be square**

```python
np.linalg.inv(A)   # but A is not square
```

Fix:

* Ensure A is (n × n)
* Or call the right function (e.g., `solve()`)

## **17.7 — Broadcasting Errors**

---

### **Error: operands could not be broadcast together**

```python
a = np.ones((3,4))
b = np.ones((3,5))
a + b    # incompatible
```

Fix:

1. Add axes explicitly:

```python
a[:, :, None]
```

2. Reshape:

```python
b.reshape(1,5)
```

3. Or repeat:

```python
np.tile(b, (3,1))
```

(Broadcasting preferred — tile is slower.)

## **17.8 — Assignment Shape Errors**

---

### **Error: could not broadcast input array from shape (…) into shape (…)**

```python
a[0:3] = np.array([1,2])  
```

Left side slice has 3 elements
Right side has only 2

Fix:

Make shapes match:

```python
a[0:3] = [1,2,3]
```

or rewrite your slice:

```python
a[0:2] = [1,2]
```
