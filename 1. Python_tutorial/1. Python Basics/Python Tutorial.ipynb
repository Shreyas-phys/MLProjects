## 2.2 Basic Slicing (start : stop : step)

Go [home](#variables) 
### Core Syntax
arr[start : stop]  
arr[start : stop : step]

### Basic Slices
```python
arr[0:3]      # [10, 20, 30], 4th element excluded!!!
arr[2:5]      # [30, 40, 50]

arr[:3]       # [10, 20, 30]   (start = 0)
arr[2:]       # [30, 40, 50]   (stop = end)
```

### Step Slices
```python
arr[::2]      # [10, 30, 50]
arr[1::2]     # [20, 40]
arr[::-1]     # [50, 40, 30, 20, 10]
arr[::-2]     # [50, 30, 10]
```
### Slicing with Negative Indices
```python
arr[-4:-1]    # [20, 30, 40]
arr[-4:]      # [20, 30, 40, 50]
arr[:-2]      # [10, 20, 30]

arr[-1:0:-1]  # [50, 40, 30, 20]
```
### Bounds Never Error (Important in ML loops)

```python
arr[0:999]    # OK
arr[-999:3]   # OK
```
Python returns what exists.

But regular indexing DOES error


```python
arr[999]      # ❌ IndexError
```

### List Methods <a id="list_methods"></a>
##### Go [home](#home) 


`````python
nums=[1,23,45,67]
# append() — add an item at the end
nums.append(10)
print(nums)

# list.pop removes and returns the last element of a list:
nums.pop()
print(nums)

# extend() — append elements from another list
# extend([20, 30]) adds two elements
nums.extend([20, 30])
print(nums)

# append([20, 30]) adds one element (a list)
nums.append([40, 50])
print(nums)

# insert(index, value) — insert at a position
nums.insert(2, 99)
print(nums)

nums.pop()
print(nums)

# pop(index) — remove and return element
nums.pop(2)
print(nums)

nums.extend([1,1,1,1,1])
print(nums)
# count() — count occurrences of a value
print(nums.count(1))  # returns 5


# remove(value) — remove first matching value
nums.remove(1)

## Searching lists:
nums.index(23)  # returns the index of the first occurrence of 23
nums.index(100)  # throws an error because 100 is not in the list
`````

# 4. Loops and List Comprehensions <a id="lists"></a>
##### Go [home](#variables)  
This section focuses on the **two most important loop tools** in Python:
- `range()`  
- `enumerate()`  

`range()` produces a sequence of integers, typically used in `for` loops.

### Syntax:
range(start, stop, step)
### Notes:
- `start` is inclusive  
- `stop` is **exclusive** (just like slicing)  
- `step` is optional (defaults to 1)

`````python
# Basic ranges
list(range(5))           # [0,1,2,3,4]
list(range(1, 6))        # [1,2,3,4,5]
list(range(2, 10, 2))    # [2,4,6,8]

# Negative step (reverse range)
list(range(10, 0, -1))   # [10,9,8,...,1]
`````

```python
for index, value in enumerate(iterable):
```
### Why it's important:
- Cleaner than using `range(len(arr))`
- Very common in ML code when looping through batches

## When to use enumerate() vs range()

### ✔ Use `enumerate()` when:
- You want **index + value**
- You care about **readability**

### ✔ Use `range()` when:
- You ONLY need indices
- You’re looping a fixed number of times (not tied to a list)


`````python
names = ["Amy", "Bob", "Charlie"]

for idx, name in enumerate(names):
    print(idx, name)

# With a custom starting index:
for idx, name in enumerate(names, start=1):
    print(idx, name)
`````

# 5. Loops and List Comprehensions <a id="list_comprehension"></a>

##### Go [home](#home) 

List comprehensions are one of the most useful Python features.
They allow you to build lists **in a single, clear, fast expression**.

General form:
```python
[new_element  for  variable in iterable   if condition]
```

## 🔥 List Comprehension Quick Syntax Reference

### 1) Transform
```python
[new for x in arr]
```
### 2) Filtering ( if statement goes at the END. ALWAYS goes after the for. It is filtering)
```python
[x for x in arr if condition]
```
### 3) Transform + Filter
```python
[new for x in arr if condition]
```
### 4) If–Else inside (goes at the START)
```python
VALUE if CONDITION else VALUE2  for x in ...

```
### 5) Nested loops (rare)
```python
[x for row in matrix for x in row]
```

✔️ If there’s an else, that whole if/else block must go BEFORE the for.

✔️ If there’s no else, the condition is a filter and must go AFTER all loops.

`````python
# They replace loops like
squares = []
for x in nums:
    squares.append(x*x)
print(squares)

#with list comprehensions, we can write this as
squares = [x*x for x in nums]
print(squares)
#The variable after for is the same variable you must use inside the expression.


# with an optional condition:
even_squares = [x*x for x in nums if x % 2 == 0] # only even numbers
print(even_squares)

#Conditional Transformation
labeled_nums = ['even' if x % 2 == 0 else 'odd' for x in nums]
print(labeled_nums)


#Nested Loops (Flattening)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print(flattened)
#Order: left-to-right = nested loops.


# with functions
def f(x):
    return x*x + 1

vals = [f(x) for x in nums]
print(vals)


#List Comprehensions + Strings
chars = [c for c in "hello"]          # ['h','e','l','l','o']
uppercased = [word.upper() for word in ["a","b","c"]]
print(chars)
print(uppercased)
`````

# 7. Strings <a id="strings"></a>
##### Go [home](#home) 


Strings are sequences
Strings can be thought of as sequences of characters. Almost everything we've seen that we can do to a list, we can also do to a string.

Such string manipulation patterns come up often in the context of data science work.
strings in Python can be defined using either single or double quotations. They are functionally equivalent.

`````python
x = 'Pluto is a planet'
y = "Pluto is a planet"
x == y
`````

Double quotes are convenient if your string contains a single quote character (e.g. representing an apostrophe).

Similarly, it's easy to create a string that contains double-quotes if you wrap it in single quotes:

```python
print("Pluto's a planet!")
print('My dog is named "Pluto"')
```
If we try to put a single quote character inside a single-quoted string, Python gets confused:

```python
'Pluto's a planet!'
```
We can fix this by "escaping" the single quote with a backslash.

```python
'Pluto\'s a planet!\n'
```
The last sequence, \n, represents the newline character. It causes Python to start a new line.

In addition, Python's triple quote syntax for strings lets us include newlines literally (i.e. by just hitting 'Enter' on our keyboard, rather than using the special '\n' sequence). We've already seen this in the docstrings we use to document our functions, but we can use them anywhere we want to define a string.

`````python
hello = "hello\nworld"
print(hello)

triplequoted_hello = """hello
world"""
print(triplequoted_hello)
triplequoted_hello == hello
`````

#### 1. Strings Are Immutable

You cannot modify a string in place.

Any "change" creates a new string object.

If you want mutability, convert to a list:

`````python
s = "hello"
s = "Hello"  # ✅ Strings are immutable, but you can reassign the variable
s[0] = "H"   # ❌ TypeError: 'str' object does not support item assignment
`````

#### 2. Strings Are Sequences

Support indexing, slicing, iteration.

`````python
s[1:4] == "ell"   # ✔
for ch in s: ...  # ✔ iterate characters
`````

#### 3. Common String Methods (All Return New Strings)

All of these create new strings; original remains unchanged.

#### 4. Strings Are Hashable

They can be used as dictionary keys or set elements.

A hashable object:

has a fixed hash value for its lifetime,

can be used as a dictionary key or a set element.

Hashability requires immutability — which is why:

strings are hashable

lists are NOT hashable

`````python
s = "alpha"
d = {s: 42}     # ✔ valid

lst = [1,2,3]
d = {lst: 42}   # ❌ TypeError: unhashable type: 'list'
`````

Why this matters (ML/physics)

Using strings as keys in config dicts / parameter dicts:
```python

```

Parsing Hamiltonian terms / labels:
```python

```

Features named by string keys (common in ML pipelines)

Rule:
✔ Strings can be dict keys because they never change → hash is stable.

### 5. Efficient String Building
❌ Slow, quadratic method:
```python
s = ""
for c in chars:
    s += c
```

This copies the whole string every loop → O(n²).

✔ Correct, fast method:
```python
s = "".join(chars)
```

.join() is optimized in C and works in linear time.

Why this matters:

If you print or log results in physics simulations

If you're parsing files line-by-line

If you're creating output strings for ML pipelines

Interview-ready statement:

“In Python, repeated concatenation is inefficient. The correct way to build large strings is using ''.join().”

### Context manager
A **context manager** is a Python construct that automatically **sets things up** and **cleans them up** for you — most commonly used with `with`.

---

#### 🚪 **Context Manager = automatic setup + automatic cleanup**

When you write:

```python
with open("data.txt", "r") as f:
    content = f.read()
```

Python does this:

1. **Open the file**
2. Give you the file object (`f`)
3. **After the block ends**, automatically:

   * closes the file
   * releases resources
   * handles exceptions safely

Even if the code crashes inside the block, the file is still closed safely.

---

#### 🧠 Why it matters?

Without a context manager, you'd have to do:

```python
f = open("data.txt", "r")
content = f.read()
f.close()   # MUST remember this manually!
```

If your code errors before `close()`, the file stays open → memory leak or locked file.

Here is a **short, clean, interview-ready File I/O cheatsheet** — perfect for quick revision and extremely relevant to physics + ML workflows.

# 11. Classes and OOP <a id="oop"></a>

Go to [Home](#home)

### ⭐ 0. Why OOP Exists (the whole point in one minute)

Imagine you’re simulating **particles**.
Without OOP, you might have:

```python
positions = [0.3, 1.1, 4.5]
velocities = [0.5, -0.2, 0.0]
masses = [1.0, 1.0, 2.0]
```

Now if you want to update each particle:

```python
for i in range(len(positions)):
    positions[i] += velocities[i] * dt
```

This is messy.
You’re manually synchronizing 3 separate lists.

**OOP solves this: bundle data + functions together.**

```python
class Particle:
    def __init__(self, x, v, m):
        self.x = x
        self.v = v
        self.m = m
    
    def update(self, dt):
        self.x += self.v * dt
```

Now each particle “knows” how to update itself:

```python
p = Particle(0.3, 0.5, 1.0)
p.update(dt)
```

**OOP = keep related data + behavior together.
Cleaner code, fewer bugs.**

---

### ⭐ 1. What is a Class? (Intuition First)

A **class** is like a template/blueprint.
An **object** is an actual thing created from that blueprint.

#### Analogy:

* Class = "Car blueprint"
* Object = "Actual car you drive"

#### Example:

```python
class Particle:
    pass
```

This class does nothing yet — but you can create objects:

```python
p1 = Particle()
p2 = Particle()
```

You now have **two separate particles**.

---

### ⭐ 2. Adding Data: `__init__` and `self`

`__init__` runs *when you create an object*.

`self` means “the object itself”.

```python
class Particle:
    def __init__(self, position, velocity):
        self.x = position
        self.v = velocity
```

Usage:

```python
p = Particle(0.0, 1.2)
print(p.x)  # 0.0
print(p.v)  # 1.2
```

**Every object gets its own x and v.**

---

### ⭐ 3. Adding Behavior: Methods

A method is just a function inside the class.

```python
class Particle:
    def __init__(self, x, v):
        self.x = x
        self.v = v
    
    def update(self, dt):
        self.x += self.v * dt
```

Now you can do:

```python
p = Particle(0.0, 2.0)
p.update(0.5)
print(p.x)  # 1.0
```

Instead of writing functions that take dictionaries or tuples, you let the object handle its own logic.

---

### ⭐ 4. Class vs Instance Variables (Super Intuitive)

#### Instance variable

Unique per object:

```python
self.x
self.v
```

#### Class variable

Shared by **all** objects:

```python
class Particle:
    count = 0  # shared!
    
    def __init__(self, x):
        self.x = x
        Particle.count += 1
```

This counts how many particles were created.

---

### ⭐ 5. `__str__` / `__repr__` (Make Objects Printable)

```python
class Particle:
    def __init__(self, x, v):
        self.x = x
        self.v = v
    
    def __repr__(self):
        return f"Particle(x={self.x}, v={self.v})"
```

Now print(p) shows:

```
Particle(x=0.3, v=1.2)
```

---

### ⭐ 6. Why Inheritance Exists (Intuition)

Imagine you have particles:

* NormalParticle
* ChargedParticle
* QuantumParticle

All particles have:

* position
* velocity
* update method

But a charged particle also has:

* charge
* force from E-field

Instead of rewriting Particle code, you extend it:

```python
class Particle:
    def __init__(self, x, v):
        self.x = x
        self.v = v

class ChargedParticle(Particle):
    def __init__(self, x, v, q):
        super().__init__(x, v)   # get parent's behavior
        self.q = q
```

Now a charged particle *is* a particle, plus more.

---

### ⭐ 7. Overriding Methods (Custom Behavior)

Parent:

```python
class Particle:
    def update(self, dt):
        self.x += self.v * dt
```

Child overrides:

```python
class ChargedParticle(Particle):
    def update(self, dt, E):
        self.v += self.q * E * dt
        self.x += self.v * dt
```

Same method name → different behavior.

This is **polymorphism**.

---

### ⭐ 8. `super()` Explained Simply

`super()` just means “use the parent version of this function”.

Example:

```python
class A:
    def greet(self):
        print("Hello from A")

class B(A):
    def greet(self):
        super().greet()   # call parent version
        print("Hello from B")
```

---

### ⭐ 9. Static and Class Methods (Clear Intuition)

#### Static method = function stored in a class

No `self`, no object.

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b
```

##### Class method = alternate constructor

Receives the class (cls), not object.

```python
class Particle:
    @classmethod
    def from_tuple(cls, tup):
        return cls(tup[0], tup[1])
```

---

### ⭐ 10. OOP Concepts in One Sentence Each

* **Encapsulation** = keep data + functions together
* **Abstraction** = hide complexity, expose simple interface
* **Inheritance** = reuse & extend code
* **Polymorphism** = same method name, different behavior

# 12. Exceptions <a id="exceptions"></a>

Here is the **cleanest, most intuitive explanation** of *why* exceptions exist, followed by a **short, sharp cheatsheet** of the essential exception patterns.

### ✔ Reason 1: Programs fail all the time

Errors happen constantly in real code:

* Missing files
* Wrong input types
* Empty arrays
* Bad indices
* Network errors
* Divide-by-zero
* Incorrect formats

Without exceptions, your program would crash immediately.

Exceptions let you **catch errors instead of crashing**, and respond gracefully.

---

### ✔ Reason 2: Exceptions separate “what you want to do” from “what might go wrong”

Normal code describes **logic**.
Exception code describes **error handling**.

Example:

```python
try:
    data = np.loadtxt("input.txt")
except FileNotFoundError:
    print("Please provide a valid data file.")
```

Instead of mixing:

* file opening
* failure cases
* recovery logic

Exceptions keep your code clean.

---

### ✔ Reason 3: Exceptions help in ML & physics more than you may think

#### **Physics/HPC use cases**

* If a data file is missing, print an error not crash.
* If dimensions mismatch, warn the user.
* Handle floating-point underflow/overflow.
* Safely close files in long simulations.
* Recover from partial simulation data.

#### **ML use cases**

* Bad hyperparameter config (e.g., negative LR)
* Mismatched shapes in matrices
* Catastrophic training failures
* Missing CSV files
* Skipping corrupt images
* Catching errors inside training loops

Exceptions keep your scripts running instead of breaking mid-HPC job.

---

### ✔ Reason 4: Exceptions give you **control**

Instead of letting Python crash, you choose what happens:

* Retry
* Skip
* Log
* Use fallback
* Fail gracefully

---

### ✔ Final intuition

> **Exceptions are for “unexpected but possible” situations.**
> They let your program survive and respond intelligently.

### 🧩 `try` / `except` — Python Interview Cheat Sheet

### **Basic pattern**

```python
try:
    risky_operation()
except SomeError:
    handle_error()
```

---

#### ✅ **Why not use `except:` (bare except)?**

A bare `except:` catches **every** exception, including things you *don’t* want to catch:

* `ZeroDivisionError`
* `TypeError`
* `ValueError`
* `KeyboardInterrupt` (Ctrl+C)
* `SystemExit`

This makes debugging difficult and hides real problems.

---

#### ✅ **Best practice: catch specific exceptions**

```python
try:
    result = a / b
except ZeroDivisionError:
    result = "error"
```

This handles the expected error and lets all others raise normally.

---

#### 🧠 **Example: safe division**

```python
def safe_div(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "error"
```

---

#### ✨ Key Takeaways

* Use **specific** exceptions (`ZeroDivisionError`) rather than `except:`.
* Only catch the errors you expect.
* `try/except` prevents crashes and lets you define safe fallback behavior.
* `finally` is used for cleanup actions.
* `with` (context managers) often removes the need for manual `try/finally`.
