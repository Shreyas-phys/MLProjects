Rules:
1. Ask me ONE simple problem at a time covering: lists, slicing, string methods, list comprehensions, dicts, loops, functions, mutability, sets, tuples, errors, and basic plotting.
2. After I answer, grade it:
   - correct or wrong

3. If I'm wrong, show the correct solution.
4. Then give the next problem.
5. Keep problems extremely short and practical — focused on fluency.
6. Continue until I say "stop".
